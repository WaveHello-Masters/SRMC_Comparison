```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.measure import marching_cubes, mesh_surface_area

# Define the function
def f(x, y, z):
    return x**2 + y**2 + z**2 - 1  # Example: Sphere level set

# Create a 3D grid of points
x = np.linspace(-1.5, 1.5, 100)
y = np.linspace(-1.5, 1.5, 100)
z = np.linspace(-1.5, 1.5, 100)
X, Y, Z = np.meshgrid(x, y, z)
values = f(X, Y, Z)  # Evaluate the function on the grid

# Extract the level set surface using marching cubes
level = 0  # Define the level set (e.g., f(x, y, z) = 0)
verts, faces, normals, _ = marching_cubes(values, level=level, spacing=(x[1] - x[0], y[1] - y[0], z[1] - z[0]))

# Visualize the extracted surface
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface using a triangular mesh
ax.plot_trisurf(verts[:, 0], verts[:, 1], faces, verts[:, 2], cmap='viridis', edgecolor='none')

# Set axis labels
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_zlabel("Z-axis")
ax.set_title("3D Level Set of f(x, y, z) = x² + y² + z² - 1")

plt.show()
```
